```python
import os

# برجاء إدخال بيانات حساب كاجل الخاص بك هنا لتشغيل الكود بنجاح
os.environ['KAGGLE_USERNAME'] = "اكتب_اسم_المستخدم_هنا"
os.environ['KAGGLE_KEY'] = "اكتب_الرمز_السري_الخاص_بك_هنا"

# الخطوة 2: استدعاء مكتبة كاجل الرسمية بعد تفعيل بيانات الحساب مباشرة
import kaggle

# الخطوة 3: تعريف قائمة تحتوي على المعرفات الرقمية والأسماء الخاصة بالملفات الطبية المستهدفة
target_datasets = [
    'mirichoi0218/insurance',
    'tboyle10/medicaltranscriptions',
    'mosapabdelghany/medical-insurance-cost-dataset',
    'chaitanyakck/medical-text',
    'harshsingh2209/medical-insurance-payout',
    'rajgupta2019/medical-insurance-dataset',
    'tarekmuhammed/patients-data-for-medical-field',
    'joebeachcapital/medical-insurance-costs'
]

print("بدء سحب وتحميل المجموعات الطبية الـ 8...\n")

# الخطوة 4: بدء الحلقة التكرارية للمرور على كل مجموعة بيانات وتحميلها تلقائياً
for dataset in target_datasets:
    
    # أ) استخراج الجزء الأخير من اسم الرابط لاستخدامه كاسم للمجلد الفرعي
    folder_name = dataset.split('/')[-1]
    
    # ب) تحديد المسار النهائي لحفظ الملفات داخل مجلد فرعي منظم يحمل اسم الداتا
    output_path = f'./medical_data/{folder_name}'
    
    print(f"جاري سحب: {folder_name} ...")
    
    # ج) محاولة تحميل الملفات وفك ضغطها تلقائياً مع معالجة الأخطاء إن وجدت
    try:
        kaggle.api.dataset_download_files(dataset, path=output_path, unzip=True)
        print(f"تم التحميل بنجاح في: {output_path}\n")
    except Exception as e:
        print(f"حدث خطأ أثناء تحميل {folder_name}: {e}\n")

print("اكتملت العملية! تم سحب جميع الملفات الطبية بنجاح.")
```

```python
import os
import pandas as pd

# تحديد المجلد الرئيسي الذي يحتوي على البيانات الطبية
base_dir = './medical_data'

# قاموس فارغ لتخزين الداتا فريمز بداخلة (اسم الداتا -> الداتا فريم الخاص بها)
medical_dataframes = {}

# المرور على المجلدات الفرعية وقراءة ملفات الـ CSV
for folder_name in os.listdir(base_dir):
    folder_path = os.path.join(base_dir, folder_name)
    
    # التأكد من أنه مجلد وليس ملفاً منفصلاً
    if os.path.isdir(folder_path):
        # البحث عن أي ملف ينتهي بـ .csv داخل المجلد
        for file in os.listdir(folder_path):
            if file.endswith('.csv'):
                file_path = os.path.join(folder_path, file)
                
                # قراءة الملف وتخزينه في القاموس باسم المجلد
                medical_dataframes[folder_name] = pd.read_csv(file_path)
                print(f"تم تحويل {file} إلى DataFrame بنجاح.")

print("\nاكتملت العملية! جميع الملفات جاهزة الآن للاستخدام.")
```

```python
import os
import pandas as pd

# تحديد المجلد الرئيسي للبيانات الطبية
base_dir = r"C:\D My Folder\medical_data"

# قاموس فارغ لتخزين جميع الداتا فريمز الـ 8
medical_dataframes = {}

# المرور على المجلدات الثمانية وقراءة المحتويات تلقائياً
for folder_name in os.listdir(base_dir):
    folder_path = os.path.join(base_dir, folder_name)
    
    # التأكد من أنه مجلد فرعي حقيقي
    if os.path.isdir(folder_path):
        for file in os.listdir(folder_path):
            file_path = os.path.join(folder_path, file)
            
            # حالة 1: إذا كان الملف CSV
            if file.endswith('.csv'):
                medical_dataframes[folder_name] = pd.read_csv(file_path)
                print(f"تم تحويل الملف [ {file} ] إلى DataFrame بنجاح.")
                break # الخروج لقراءة المجلد التالي
                
            # حالة 2: إذا كان الملف إكسل Excel
            elif file.endswith('.xlsx') or file.endswith('.xls'):
                medical_dataframes[folder_name] = pd.read_excel(file_path)
                print(f"تم تحويل ملف الإكسل [ {file} ] إلى DataFrame بنجاح.")
                break
                
            # حالة 3: إذا كان الملف نصي Text (مثل ملف تفريغ التقارير الطبية)
            elif file.endswith('.txt'):
                # قراءة الملفات النصية كأعمدة منظمة
                medical_dataframes[folder_name] = pd.read_csv(file_path, sep="\t", header=None, names=["Medical_Text"])
                print(f"تم تحويل الملف النصي [ {file} ] إلى DataFrame بنجاح.")
                break

print("\nاكتملت العملية! جميع الملفات الـ 8 تم تحويلها إلى داتا فريمز وجاهزة للاستخدام.")
```
